#Variáveis são usadas para armazenar informações.Neste exemplo, criamos variáveis para armazenar o nome, a idade e a altura de uma pessoa e, em seguida, usamos a função print para exibir esses valores.

#Este exercício solicita ao usuário que insira dois números, calcula a soma desses números e exibe o resultado. A função input é usada para obter a entrada do usuário, e float converte a entrada para um número decimal.

```python
numero1 = float(input("Digite o primeiro número: "))
numero2 = float(input("Digite o segundo número: "))

soma = numero1 + numero2

print(f"A soma de {numero1} e {numero2} é {soma}")
```

#Este exercício verifica se um número é par ou ímpar. A operação % (módulo) retorna o resto da divisão do número por 2. Se o resto for 0, o número é par; caso contrário, é ímpar.

```python
numero =int(input("Digite um numero inteiro"))

if numero % 2 == 0:
  print(f"O {numero} é par.")
else:
  print(f"O {numero} é impar.")
```
